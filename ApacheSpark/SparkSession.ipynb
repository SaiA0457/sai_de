## Lazy evaluation and actions
All transformations are planned but not executed until an action is triggered. When an action is performed, Spark executes all pending transformations in an optimized manner.

Common actions available in Spark DataFrame API :

- **collect()**: Retrieves all rows as a list to the driver.

  Example: rows = df.collect()

- **count()**: Returns the number of rows in the DataFrame.

  Example: total = df.count()

- **show()**: Displays the top rows in tabular format (use display() in Databricks).

  Example: df.show()

- **display()**: Renders DataFrame in Databricks notebook UI.

  Example: display(df)

- **take(n)**: Returns the first n rows as a list.

  Example: first_two = df.take(2)

- **first()**: Returns the first row.

  Example: row = df.first()

- **head()**: Returns the first n rows (default n=1).

  Example: top_rows = df.head(3)

- **foreach()**: Applies a function to each row (no return).

  Example: df.foreach(lambda row: print(row))

- **foreachPartition()**: Applies a function to each partition.

  Example: df.foreachPartition(lambda partition: print(list(partition)))

- **reduce()**: Aggregates rows using a function.

  Example: result = df.rdd.reduce(lambda x, y: x + y)

- **saveAsTable()**: Saves DataFrame as a table in metastore.

  Example: df.write.saveAsTable("my_table")

- **write()**: Writes DataFrame to external storage (e.g., parquet, csv).

  Example: df.write.parquet("/path/to/output")

- **toPandas()**: Converts DataFrame to Pandas DataFrame.

  Example: pdf = df.toPandas()

- **explain()**: Prints the execution plan for the DataFrame.

  Example: df.explain()
