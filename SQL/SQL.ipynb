```python
%sql

create table customers(
  customer_id integer,
  customer_name varchar(20),
  city varchar(20));

insert into customers (customer_id, customer_name, city) values
  (1, 'Ravi', 'Hyderabad'),
  (2, 'Ankit', 'Delhi'),
  (3, 'Priya', 'Bangalore'),
  (4, 'John', 'Mumbai');
```

```python
%sql
create table orders(
  order_id integer,
  customer_id integer,
  order_date date,
  amount integer);

insert into orders (order_id, customer_id, order_date, amount) values
  (101, 1, '2024-01-01', 500),
  (102, 1, '2024-01-03', 700),
  (103, 2, '2024-01-04', 300),
  (104, 2, '2024-01-06', 900),
  (105, 1, '2024-01-10', 200)
```

```python
%sql

select c.customer_name, o.order_id, o.amount from customers as c left join orders as o on c.customer_id = o.customer_id
```

```python
%sql
select c.customer_id, c.customer_name, count(o.order_id) from customers as c left join orders as o on c.customer_id = o.customer_id
group by c.customer_id, c.customer_name
order by c.customer_id
```

```python
%sql

select c.customer_id, c.customer_name from customers as c left join orders as o on c.customer_id = o.customer_id
where order_id is null
```

```python
%sql
select c.customer_id, sum(o.amount) as total_spent
from customers c
left join orders o on c.customer_id = o.customer_id
group by c.customer_id
order by total_spent desc
limit 1


```

```python
%sql
select c.customer_id, round(avg(o.amount),2) as avg_order_amount
from customers c
left join orders o on c.customer_id = o.customer_id
group by c.customer_id
```

```python
%sql
select c.customer_id, c.customer_name, count(o.order_id) from customers as c left join orders as o on c.customer_id = o.customer_id
group by c.customer_id, c.customer_name
having count(o.order_id) > 2
order by c.customer_id;
```

```python
%sql
select o.order_id, c.customer_name, o.amount from customers as c inner join orders as o on c.customer_id = o.customer_id
where c.city = 'Hyderabad'
```

```python
%sql
select c.customer_name, sum(o.amount) as total_spent from customers as c inner join orders as o on c.customer_id = o.customer_id
group by c.customer_name
having total_spent> '1000'
```

```python
%sql
select customer_name, order_date from (
  select *, 
    row_number() OVER(partition by c.customer_id order by o.order_date desc) as rn
  from customers as c left join orders as o on c.customer_id = o.customer_id
)
where rn = 1;
```
